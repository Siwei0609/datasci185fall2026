````python
# Write your answer
pd_df= pd.read_csv("data/korea_emission.csv")
plt.scatter(x= pd_df["Energy_consumption"] , y=pd_df["CO2_emission"])
plt.xlabel("Energy_consumption")
plt.ylabel("CO2_emission")
plt.title("Energy_consumption vs. CO2_emission_in_korea")
plt.show()

````

## (Q6) Choose any 2 of the countries in the "data" subfolder and import the 2 datasets as you did in (Q2). Create two separate scatter plots that plot "Year" vs. "CO2 emissions".

- For full points, label the axes in both plots. The labels should clearly indicate what variables are being plotted.

- Use `plt.title` to create a title for both plots. The titles should specify which country the data corresponds to and what information is being displayed.

````python
china_emission = pd.read_csv("data/china_emission.csv")
japan_emission = pd.read_csv("data/japan_emission.csv")

plt.scatter(x=china_emission["Year"], y=china_emission["CO2_emission"])
plt.xlabel("Year")
plt.ylabel("CO2 Emission")
plt.title("Year vs. CO2 Emission in China")
plt.show()

plt.scatter( x=japan_emission["Year"], y=japan_emission["CO2_emission"])
plt.xlabel("Year")
plt.ylabel("CO2 Emission")
plt.title("Year vs. CO2 Emission in Japan")
plt.show()
````
